```python
!pip install langchain openai
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
```

```python
# Task 1
template_1 = PromptTemplate(input_variables=['topic'], template='Explain {topic} in simple terms for beginners')
print(template_1.format(topic='Artificial Intelligence'))
```

```python
# Task 2
template_2 = PromptTemplate(input_variables=['topic','audience','tone'], template='Explain {topic} for {audience} in a {tone} tone')
print(template_2.format(topic='AI', audience='beginners', tone='friendly'))
print(template_2.format(topic='Python', audience='kids', tone='fun'))
print(template_2.format(topic='Deep Learning', audience='engineers', tone='technical'))
```

```python
# Task 3
teaching_template = PromptTemplate(input_variables=['topic'], template='Explain {topic} clearly step by step')
interview_template = PromptTemplate(input_variables=['topic'], template='Ask 3 interview questions about {topic}')
story_template = PromptTemplate(input_variables=['topic'], template='Explain {topic} as a story')
topic = 'Machine Learning'
print(teaching_template.format(topic=topic))
print(interview_template.format(topic=topic))
print(story_template.format(topic=topic))
```

```python
# Task 4
system_template = SystemMessagePromptTemplate.from_template('You are a helpful {role}')
human_template = HumanMessagePromptTemplate.from_template('Explain {topic}')
chat_prompt = ChatPromptTemplate.from_messages([system_template, human_template])
messages = chat_prompt.format_messages(role='teacher', topic='Neural Networks')
for msg in messages:
    print(msg.content)
```

```python
# Task 5
def validate_inputs(audience, tone):
    valid_audience = ['beginner','intermediate','expert']
    valid_tone = ['formal','casual','fun']
    if audience not in valid_audience:
        print('Invalid audience! Defaulting to beginner')
        audience = 'beginner'
    if tone not in valid_tone:
        print('Invalid tone! Defaulting to formal')
        tone = 'formal'
    return audience, tone
print(validate_inputs('kids','funny'))
```

```python
# Task 6
def generate_prompt(topic, audience, tone, style):
    audience, tone = validate_inputs(audience, tone)
    styles = {
        'teaching': 'Explain {topic} step by step for {audience} in a {tone} tone',
        'interview': 'Ask questions about {topic} for {audience} in a {tone} tone',
        'storytelling': 'Explain {topic} as a story for {audience} in a {tone} tone'
    }
    template = PromptTemplate(input_variables=['topic','audience','tone'], template=styles[style])
    return template.format(topic=topic, audience=audience, tone=tone)
print(generate_prompt('Neural Networks','beginner','fun','storytelling'))
```

```python
# Task 7
reusable_template = PromptTemplate(input_variables=['topic'], template='Explain {topic} in simple terms')
topics = ['AI','Python','Blockchain','Cloud Computing','Cyber Security']
for t in topics:
    print(reusable_template.format(topic=t))
```

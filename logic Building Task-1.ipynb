```python
# 1. User Login Check

print("-----Q1: User Login Check -----")

# Stored credentials
stored_username = "admin"
stored_password = "1234"

# Given credentials (you can change to test)
username = "admin"
password = "1234"

# Login validation
if username == stored_username and password == stored_password:
    print("Login Successful")
else:
    print("Invalid Credentials")

```

```python
# 2. Pass / Fail Analyzer


print("\n----- Q2: Pass / Fail Analyzer -----")

marks = [45, 78, 90, 33, 60]

pass_count = 0
fail_count = 0

for mark in marks:
    if mark >= 50:
        pass_count += 1
    else:
        fail_count += 1

print("Total Students:", len(marks))
print("Total Passed Students:", pass_count)
print("Total Failed Students:", fail_count)

```

```python
# 3. Simple Data Cleaner


print("\n----- Q3: Simple Data Cleaner -----")

names = [" Alice ", "bob", " CHARLIE "]

cleaned_names = []

for name in names:
    cleaned_name = name.strip().lower()
    cleaned_names.append(cleaned_name)

print("Original Names:", names)
print("Cleaned Names:", cleaned_names)
```

```python
# 4. Message Length Analyzer


print("\n----- Q4: Message Length Analyzer -----")

messages = ["Hi", "Welcome to the platform", "OK"]

for message in messages:
    length = len(message)
    print(f"Message: '{message}' | Length: {length}")
    
    if length > 10:
        print("Flag: Message longer than 10 characters")
```

```python
# 5. Error Message Detector


print("\n----- Q5: Error Message Detector -----")

logs = ["INFO", "ERROR", "WARNING", "ERROR"]

error_count = logs.count("ERROR")

print("Total ERROR entries:", error_count)
```
